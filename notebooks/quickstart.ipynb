# 🧠 NeuroSpike: Neuromorphic Computing Quick Start

Welcome to **NeuroSpike**, a high-performance Spiking Neural Network (SNN) library supporting:
- Event-driven vision and temporal audio datasets (Spiking Heidelberg Digits, DVS-CIFAR10)
- Biologically inspired neuron models (Adaptive LIF, Izhikevich, Stochastic LIF)
- Spike-driven Transformers and ResNets
- Unsupervised STDP synaptic plasticity
- Hardware energy efficiency benchmarking (SynOps vs MACs)

```python
import torch
from neurospike.models import SpikingTransformer, SpikingCNN, AdaptiveLIF, IzhikevichNeuron
from neurospike.data import RateCoding, TemporalCoding, SHDDataset
from neurospike.visualization import plot_spike_raster, plot_membrane_potential, plot_stdp_window

print("NeuroSpike imported successfully!")
```

## 1. Spike Encoding: Analog Signal to Spike Train

```python
# Continuous sensory signal
signal = torch.rand(1, 32)
encoder = RateCoding(num_steps=50, gain=1.0)
spikes = encoder(signal)
print(f"Spike train shape (T, B, features): {spikes.shape}")
print(f"Average firing rate: {spikes.mean().item():.3f}")
```

## 2. Advanced Neuron Dynamics: Adaptive LIF & Izhikevich Models

```python
# Instantiate intrinsically bursting Izhikevich neuron
neuron = IzhikevichNeuron.from_pattern("intrinsically_bursting")
v, u = neuron.init_state(1)
current = torch.tensor([12.0])

v_trace = []
for t in range(100):
    spk, v, u = neuron(current, v, u)
    v_trace.append(v.item())

print(f"Simulated 100 timesteps of biological membrane potential.")
```

## 3. Spike-Driven Transformer on Neuromorphic Audio

```python
# Spiking Heidelberg Digits model
shd_transformer = SpikingTransformer(
    input_dim=700,
    num_classes=20,
    d_model=128,
    num_heads=4,
    num_layers=2,
    num_steps=20
)

# Sample audio event stream: (B=2, T=20, Channels=700)
audio_events = (torch.rand(2, 20, 700) < 0.05).float()
logits = shd_transformer(audio_events)
print(f"Classification output logits: {logits.shape}")
```
